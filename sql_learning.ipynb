
## MySQL Categories

1) DDL (Data Definition Language):

Purpose: Define or modify the structure of the database (e.g., schemas, tables, indexes).

- CREATE (to create databases, tables, indexes, etc.)

- ALTER (to modify an existing database object)

- DROP (to delete objects)

- TRUNCATE (to remove all records from a table)

- RENAME (to rename an object)


==================================


2) DML (Data Manipulation Language):

Purpose: Manipulate the data stored in the database.


- SELECT (to query data)

- INSERT (to add new records)

- UPDATE (to modify existing records)

- DELETE (to remove records)

========================================

3) DCL (Data Control Language):

Purpose: Control access to data and database objects.

- GRANT (to give user access privileges)

- REVOKE (to take away privileges)


============================================

4) TCL (Transaction Control Language):

Purpose: Manage transactions within the database.


- COMMIT (to save transactions permanently)

- ROLLBACK (to undo transactions)

- SAVEPOINT (to set a point within a transaction to which you can roll back)

Syntax: 
1) *CREATE TABLE table_name (column1 datatype constraint, column2 datatype constraint....);*

**`WITH` (CTE) vs `TEMPORARY TABLE`**

| `WITH` (CTE) | `TEMPORARY TABLE` |
|--------------|-------------------|
| Temporary result set | Temporary table |
| Exists for **one SQL statement** | Exists for the **entire session** |
| Cannot be reused after the query | Can be reused across multiple queries |
| Usually read-only | Can `INSERT`, `UPDATE`, `DELETE` |
| Best for simplifying complex queries | Best for multi-step data processing |

**Rule of Thumb**
- **`WITH`** → Use for a temporary result **within a single query**.
- **`TEMPORARY TABLE`** → Use when the data needs to **persist across multiple queries** during the session.

4) Foreign Key Constraint

- Scenario: One Foreign Key (Driver to Ride Relationship)
- We’ll now establish a one-to-many relationship between `drivers` and `rides`, where one driver can have many rides.

````python
%%sql
DROP TABLE IF EXISTS UBER.rides;

DROP TABLE IF EXISTS UBER.drivers;
CREATE TABLE UBER.drivers (
    driver_id INT PRIMARY KEY,
    driver_name VARCHAR(100),
    license_number VARCHAR(50) UNIQUE
);

-- Creating the rides table with driver_id as a foreign key
CREATE TABLE UBER.rides (
    ride_id INT PRIMARY KEY,
    driver_id INT,
    pickup_location VARCHAR(100),
    dropoff_location VARCHAR(100),
    ride_date DATETIME,
    fare DECIMAL(10, 2),
    FOREIGN KEY (driver_id) REFERENCES drivers(driver_id)
);

````

    FOREIGN KEY (driver_id) REFERENCES drivers(driver_id) - What we mean here is, pick "driver_id" from the "UBER" Database and use it in rides table. 

    - The core concept is using primary key from one table to another as foreign key

    "driver_id" is marked as "MUL" in the key section which means it is being used as "Foreign Key" in some other table.

````python
%%sql
DROP TABLE IF EXISTS UBER.rides;

DROP TABLE IF EXISTS UBER.drivers;
CREATE TABLE UBER.drivers (
    driver_id INT PRIMARY KEY,
    driver_name VARCHAR(100),
    license_number VARCHAR(50) UNIQUE
);

-- Creating the rides table with driver_id as a foreign key
CREATE TABLE UBER.rides (
    ride_id INT PRIMARY KEY,
    driver_id INT,
    pickup_location VARCHAR(100),
    dropoff_location VARCHAR(100),
    ride_date DATETIME,
    fare DECIMAL(10, 2),
    FOREIGN KEY (driver_id) REFERENCES drivers(driver_id) ON DELETE CASCADE
);

````

The keyword "ON DELETE CASCADE" means npow that we can delete a primary key "drivers_id" from "drivers" table first, but as a Cascading effect the records of "rides" that used the respective "drivers_id" shall be lost.

Now let's say:

- I do not want cascading effect, and I dont want to permenantly delete a ride record from the "rides" table, but i do not want it to be considered for analysis.
- Yet, the rides table allow me to delete records as it has no foreign key anywhere else, in some circumstances it might get deleted.
- We do not want this, hence do something called "Soft Delete", where we add a new column to the rides table as "Is_delete" and set it as Boolean with "False" as default value, and set it to "True" if that record needs to be deleted, but not permanantly delete.

- We could simply add ride_id to drivers table as foreign key to prevent deletion from the rides table
- But then I would have multiple number of same driver's records with multiple "ride_id" which causes redundancy.
- Hence soft delete prevent both from deletion and from redundancy.

````python
%%sql 

ALTER TABLE rides ADD COLUMN Is_Delete BOOLEAN DEFAULT false;
````

    Here that we have used "Default" Keyword which provides a default value for the respective column even if no data is provided at the time of execution. And it overwrites the default if a new value other than default is provided.

````python
%%sql

UPDATE rides SET Is_Delete = true WHERE driver_id = 101;
````

````python
%%sql

SELECT * FROM rides;
````

    This way Analyst omit the record with Is_Delete = 0 and consider it as deleted record but the actual data might not be deleted.

5) MINIMAL

In this context, minimal means that the set of columns (attributes) chosen to uniquely identify a row contains no unnecessary parts—if you remove any column from that set, 
it would no longer uniquely identify the row.

Breaking It Down with an Example
Imagine a table with columns A, B, and C where:

The combination (A, B, C) uniquely identifies each row.
However, if it turns out that (A, B) alone is enough to uniquely identify each row, then (A, B, C) is not minimal because C is extra.
Thus, the minimal key here is (A, B) since you cannot remove A or B without losing the ability to uniquely identify the row. This minimal set is what we call a candidate key.

One-Line Definition Recap
Minimal (in this context): A set of columns that is just sufficient to uniquely identify a row—removing any column would break this uniqueness.

````python
%%sql

CREATE TABLE ENROLLMENT (
    student_id INT NOT NULL,
    course_id INT NOT NULL,
    enrollment_date DATE,
    course_name VARCHAR(100),
    PRIMARY KEY ( student_id, course_id)
)
````

Now in the above query:
- We can use student_id and course_id to uniquely identify a record, but if I remove any one of them, automatically the uniqueness would vanish.
- That's called MINIMAL. But let's say if I add "course_name" to primary key list (it'll brecome not minimal), but even without course_name we can uniquely identify the records.
- MINIMAL is the minimum required unique columns to identify a record.

    Example 2:
    If I use "email" as a primary key, it is unique by its own, but in order to improve the uniqueness I can add "Adhaar Number" and "Phone Number", Now all these combined (Composite Primary Key) are not "Minimal", but now If I remove Adhaar and Phone Number, "Email" alone is enough to identify records unqiuely.

1. Natural Key
A natural key is a real-world, meaningful piece of data (or combination of columns) that already uniquely identifies a record.

Examples:

A Social Security Number (SSN) in the U.S. (if each person has a unique SSN).
An email address if each user must have a unique email.
A product serial number that is truly unique for each product.

2. Surrogate Key
A surrogate key is a synthetic or artificial key added to a table just to uniquely identify each row. It typically has no meaning in the real world; it’s there only for the database’s sake.

Examples:

An auto-increment integer column like id.
A UUID (a universally unique identifier) column.
Any system-generated unique value.



Summary 
Natural Key: Real data with business meaning, used as an identifier.
Surrogate Key: Artificial (often numeric) identifier, stable and system-generated.

Example 

    CREATE TABLE dim_customer (
        customer_key INT AUTO_INCREMENT ,         -- Surrogate Key (unique for each version of the customer record)
        aadhar_number INT NOT NULL PRIMARY KEY,   -- Natural Key (unique identifier for the customer)
        customer_name VARCHAR(100),
        city VARCHAR(100),
        effective_date DATE,              -- Date this record version became active
        end_date DATE,                    -- Date this record version ended
        is_current BOOLEAN                -- Flag to indicate the current active record
);

1) Natural Key is naturally unique across all aspects ( Ex: Adhaar Number is unique for all citizen regardless of which brands customer he/she is).
- All natural keys are capable of being primary key, regardless we still set natural keys with constraint as primary key, as when data is being manually entered, a chance of error or duplication exists.

Here's the breakdown:

**Super Key**
Any set of one or more attributes that can uniquely identify a row in a table. It can contain extra, unnecessary attributes.

- Example: In a `Students` table with (RollNo, Name, Email), the sets {RollNo}, {RollNo, Name}, {RollNo, Email}, {RollNo, Name, Email} are all super keys — because each one uniquely identifies a row (even though some carry redundant columns).

**Candidate Key**
A *minimal* super key — no attribute can be removed without losing uniqueness. It's a super key with no redundancy.

- From the example above, {RollNo} alone is a candidate key (assuming RollNo alone is unique). {RollNo, Name} is NOT a candidate key because Name is unnecessary — removing it still keeps uniqueness.
- A table can have multiple candidate keys (e.g., both RollNo and Email might independently be unique identifiers).

**Primary Key**
One candidate key chosen by the database designer to be *the* main identifier for the table. Only one primary key per table, and it cannot contain NULLs.

- If both RollNo and Email are candidate keys, you pick one (say RollNo) as the primary key. The other (Email) becomes an "alternate key."

**Composite Primary Key**
A primary key made up of **two or more attributes together**, where no single attribute alone is unique enough, but the combination is.

- Example: In an `Enrollments` table tracking which students take which courses, no single column (StudentID or CourseID) is unique on its own — a student can enroll in many courses, and a course has many students. But the pair (StudentID, CourseID) together uniquely identifies each enrollment row. This combined key is the composite primary key.

**Quick summary table**

| Concept | Definition | Minimal? | Can be multiple per table? |
|---|---|---|---|
| Super Key | Any attribute set that uniquely identifies rows | No | Yes |
| Candidate Key | Minimal super key | Yes | Yes |
| Primary Key | The chosen candidate key | Yes | No (only 1) |
| Composite Primary Key | Primary key made of 2+ attributes | Yes (as a combination) | No (only 1, but multi-column) |

**Relationship:** Every candidate key is a super key, every primary key is a candidate key, and a composite primary key is just a primary key that happens to need more than one column.

Okay, imagine a classroom full of kids, and you want a way to call on one specific kid without pointing.

**Super Key** = Any way of describing a kid that works, even if you use more words than you need.
- "The kid wearing a red shirt, sitting in seat 5, named Sam" — that works, but it's way more than you need.

**Candidate Key** = The *shortest* way of describing a kid that still works — you can't remove anything else without getting confused about who you mean.
- "Seat 5" — that's enough by itself. Nothing extra.
- Maybe "Full Name + Birthday" also works by itself, as a different shortest way.
- So a class might have a few different "shortest ways" — those are all candidate keys.

**Primary Key** = The teacher picks ONE of those shortest ways to always use, so everyone agrees.
- The teacher says: "From now on, I'll always call kids by their Seat Number."

**Composite Primary Key** = Sometimes ONE thing isn't enough to point at just one kid, so you need TWO things together.
- Imagine instead of a classroom, it's a big lunchroom with tables, and lots of kids share the same seat number at different tables. Seat number alone isn't enough anymore!
- But "Table Number + Seat Number" together always points to exactly one kid. That combo is the composite primary key.

**Tiny recap:**
- Super key = any description that works (even a wordy one) 🗣️
- Candidate key = the shortest working description ✂️
- Primary key = the one shortest description you officially choose 🏆
- Composite primary key = when you need two things *together* to point at one kid 🤝

COALESCE is a SQL function that returns the **first non-NULL value** from a list of expressions you give it. If every expression turns out to be NULL, it returns NULL.

## Syntax

```sql
COALESCE(expr1, expr2, expr3, ..., exprN)
```

It evaluates the arguments **in order**, left to right, and stops as soon as it finds one that isn't NULL.

## How it works, step by step

```sql
SELECT COALESCE(NULL, NULL, 5, 10);
-- Result: 5
```

- `NULL` → skip
- `NULL` → skip
- `5` → not NULL, return it, stop here (never even looks at `10`)

## Common use cases

**1. Replacing NULLs with a default value**
```sql
SELECT COALESCE(discount, 0) AS discount
FROM orders;
```
If `discount` is NULL, this shows `0` instead.

**2. Falling back across multiple columns**
```sql
SELECT COALESCE(mobile_phone, home_phone, work_phone, 'No phone on file') AS contact_number
FROM customers;
```
Tries mobile first, then home, then work, then finally a text fallback.

**3. Combining with calculations so NULL doesn't wipe out a result**
```sql
SELECT price * COALESCE(quantity, 1) AS total
FROM items;
```
Without COALESCE, `price * NULL` would just give NULL for the whole row.

## Key details worth knowing

- **All arguments should be a compatible type** — the function will try to convert them to a common type, but wildly mismatched types (like text and date) can cause an error.
- **It's essentially shorthand for a CASE expression:**
  ```sql
  COALESCE(a, b, c)
  -- is equivalent to:
  CASE
    WHEN a IS NOT NULL THEN a
    WHEN b IS NOT NULL THEN b
    ELSE c
  END
  ```
- **It's ANSI SQL standard**, so it works consistently across PostgreSQL, MySQL, SQL Server, Oracle, SQLite, etc. (unlike vendor-specific alternatives like `ISNULL()` in SQL Server or `NVL()` in Oracle, which only take two arguments).
- **Short-circuits evaluation** — this matters if a later argument is an expensive computation or a subquery, since it won't run unless needed.

## Subquery — Quick Reference

A subquery is a query nested inside another query, always enclosed in **parentheses**.

```sql
SELECT column1
FROM table1
WHERE column2 = (SELECT column2 FROM table2 WHERE condition);
```

---

### 1. Where Subqueries Can Be Placed

| Placement | Example |
|---|---|
| **SELECT clause** | `SELECT name, (SELECT MAX(amount) FROM orders) AS max_order FROM customers;` |
| **FROM clause** (derived table) | `SELECT * FROM (SELECT customer_id, SUM(order_amount) AS total FROM orders GROUP BY customer_id) AS t;` |
| **WHERE clause** | `SELECT * FROM customers WHERE customer_id IN (SELECT customer_id FROM orders WHERE order_amount > 3000);` |
| **HAVING clause** | `SELECT customer_id, SUM(order_amount) FROM orders GROUP BY customer_id HAVING SUM(order_amount) > (SELECT AVG(order_amount) FROM orders);` |
| **JOIN clause** | `SELECT * FROM customers c JOIN (SELECT customer_id, COUNT(*) cnt FROM orders GROUP BY customer_id) o ON c.customer_id = o.customer_id;` |

---

### 2. Types by Result Shape

| Type | Returns | Use with |
|---|---|---|
| **Scalar** | 1 row, 1 column | `=`, `>`, `<`, etc. |
| **Row** | 1 row, multiple columns | `=`, `IN` (with row constructor) |
| **Column/Table** | multiple rows, 1 column | `IN`, `ANY`, `ALL` |
| **Correlated** | re-evaluated per outer row | `EXISTS`, `NOT EXISTS`, or referencing outer table |

---

### 3. Core Syntax Rules

- **Always wrap in `()`** — non-negotiable.
- **Scalar subquery** (used with `=`, `>`, etc.) must return **exactly one row, one column** — else runtime error.
- **`IN` / `NOT IN` / `ANY` / `ALL`** subqueries can return multiple rows, but only **one column**.
- **FROM-clause subqueries (derived tables) must have an alias** — MySQL/most DBs will error without one.
  ```sql
  SELECT * FROM (SELECT ...) AS alias_name;  -- alias mandatory
  ```
- **`ORDER BY` inside a subquery is generally meaningless** unless paired with `LIMIT`/`TOP` — most optimizers ignore it otherwise.
- **No `AS` needed for the subquery itself** in `WHERE`/scalar context, only when it's a derived table in `FROM`/`JOIN`.

---

### 4. Correlated vs Non-Correlated

**Non-correlated** — runs once, independent of outer query:
```sql
SELECT * FROM customers 
WHERE customer_id IN (SELECT customer_id FROM orders);
```

**Correlated** — references outer query column, runs once *per outer row* (slower):
```sql
SELECT c.customer_name
FROM customers c
WHERE EXISTS (
    SELECT 1 FROM orders o 
    WHERE o.customer_id = c.customer_id AND o.order_amount > 3000
);
```

---

### 5. Key Rules / Gotchas

1. **`NOT IN` + NULLs = danger.** If the subquery result contains even one `NULL`, `NOT IN` returns **no rows at all** (silently). Prefer `NOT EXISTS` instead.
2. **Use `EXISTS`/`NOT EXISTS` for correlated existence checks** — usually faster than `IN` for large datasets since it can short-circuit.
3. **Scalar subqueries returning >1 row** → runtime error ("subquery returns more than 1 row").
4. **Can't use `ORDER BY` on the outer query referencing a derived table column** unless that column is explicitly selected out.
5. **Subqueries in `SELECT`** run once per outer row too (like correlated) — expensive if unindexed.
6. **Aggregate in subquery, no `GROUP BY` needed** if you want one overall value (e.g., `SELECT AVG(x) FROM t` returns scalar automatically).
7. **Can be nested multiple levels deep**, but readability drops fast — consider CTEs (`WITH`) instead beyond 2 levels.

---

### 6. Quick Decision Guide

| Need | Use |
|---|---|
| Compare to a single aggregate value | Scalar subquery |
| Check membership in a list | `IN (subquery)` |
| Check existence, correlated to outer row | `EXISTS (subquery)` |
| Reusable computed table | Derived table in `FROM` or a CTE (`WITH`) |
| Filter groups by aggregate condition | Subquery in `HAVING` |

````python
%%sql


CREATE TABLE Sales (
    TransactionID INT,
    Store VARCHAR(50),
    SalesAmount DECIMAL(10, 2)
);

INSERT INTO Sales (TransactionID, Store, SalesAmount)
VALUES
    (1, 'A', 100.00),
    (2, 'A', 200.00),
    (3, 'A', 150.00),
    (4, 'B', 250.00),
    (5, 'B', 300.00);
````

````python
%%sql

SELECT Store, SUM(SalesAmount) FROM Sales GROUP BY Store;
````

    Here that we have made the aggregation but the number of records in the initialy table is reduced to 2 records.

    Now lets say the owner of the store asks for the total amount sales done in both the shops, you casn easily aggregate the amount by and use GROUP BY to group them, but now if he/she asks the data along with "Transaction_id", You can do the following:

````python
%%sql

SELECT TransactionID,Store, SUM(SalesAmount) FROM Sales GROUP BY Store, TransactionID;
````

    Now you can see, that the GROUP BY has happened yet still, the amounts are not aggregated as the TransactionIDs are unique.
    

````python
%%sql

SELECT TransactionID, Store, SalesAmount,
       SUM(SalesAmount) OVER (PARTITION BY Store ) AS TotalSales
FROM Sales;
````

    Here you can see that the resultant table's rows is not reduced, it is same as the original table, and the transactionIDs also exist for all the records.

    If they keyword "OVER" exists next to an Aggregator function then the given query is a window function,

    "OVER" takes 2 arguments along with only 2 keywords ("PARTITION BY", "ORDER BY")

    Syntax: OVER( PARTITION BY Column 1 ORDER BY Column 2).

    In the above query, the the store is partitioned into all the distinct stores (A and B), and then SUM aggregator is used individually for each store records. Now if I add ORDER BY and provide a column as argument the records shall be order by that column.

````python
%%sql

SELECT TransactionID, Store, SalesAmount,
       SUM(SalesAmount) OVER (PARTITION BY Store ORDER BY TransactionID ) AS TotalSales
FROM Sales;
````

    It has ordered the records by transactionId now, Lets try descending order

````python
%%sql

SELECT TransactionID, Store, SalesAmount,
       SUM(SalesAmount) OVER (PARTITION BY Store  ORDER BY TransactionID DESC ) AS TotalSales
FROM Sales;
````

    Now it has ordered in descending order, as you can see, 3,2,1 or in descending then 5,4. Here the descending ordering is fiorst applied to the 1 st partition (A) then to (B), hence the result. That is, All the A records are in a window, and all the B records are in different window.

![Window Functions](window_functions.png)


A window function computes a value across a set of rows related to the current row — but unlike `GROUP BY`, it doesn't collapse those rows into one. Each row keeps its own identity and gets an extra calculated column alongside it.

**The logic:**
1. `PARTITION BY` splits the rows into groups (like `GROUP BY`, but nothing collapses).
2. `ORDER BY` (inside `OVER`) decides the sequence within each group — needed for things like running totals or "previous row" lookups.
3. The function then computes its result per row, looking at that row's partition/frame.

**Example:**

```sql
SELECT
  employee,
  department,
  salary,
  AVG(salary) OVER (PARTITION BY department) AS dept_avg
FROM employees;
```

| employee | department | salary | dept_avg |
|---|---|---|---|
| Alice | Sales | 50000 | 55000 |
| Bob | Sales | 60000 | 55000 |
| Carol | IT | 70000 | 70000 |

Each row still shows its own salary, but `dept_avg` is computed by looking at all rows in that employee's department — Alice and Bob both see 55000 (the Sales average), while a plain `GROUP BY` query would have given you only one row per department and lost Alice's and Bob's individual salaries.

That's the core logic: **partition the rows, look across the group without merging them, attach the result to every row that belongs to it.**

    Aggregators, Rankings, Values all of these are windows function if OVER is used next to them. For all the Aggregator functions, the syntax is same as above replacing "SUM()" with any function

    P0(Partition 0)                                P1(Partition 1)

    (104, '2025-01-01 10:15:00', 25.00),           (103, '2025-01-01 10:10:00', 22.75),


    And it continues to take reminders of each sensor_id by 2 and according to the reminder it'll put them into a partition in spite the snesor_id itself being unique.

    You might ask, but there is no grouping here, yes is there none, but rather than scaning entire table, we have successfully partitioned a table with basically no grouping property (all are columns with unique attributes) into "n" partitions, this will reduce us from half the uncessary records.

## How MySQL partition indexes actually work

MySQL only supports **local indexes**, not global indexes (unlike Oracle, which lets you choose). This means:

- Every index you define on a partitioned table is **automatically and identically applied to every partition and subpartition**.
- You can't create an index that exists on only *some* partitions, and you can't create *different* indexes per partition.
- Physically, though, each partition/subpartition **does** get its own separate copy of the index structure (its own B-tree). So in a sense, you already *do* have "individual indexes per partition" — MySQL just manages them as one logical index applied uniformly.

Example:

```sql
ALTER TABLE orders ADD INDEX idx_customer (customer_name);
```

This creates the `idx_customer` index **inside every one of your 6 subpartitions** (3 partitions × 2 subpartitions each) — not one global index spanning the whole table. When you query with a `WHERE order_date = ... AND customer_name = ...`, MySQL prunes to the relevant partition first, then uses that partition's local copy of the index.

## What you *can't* do

```sql
-- NOT possible: index only on p_2020, not on others
ALTER TABLE orders ADD INDEX idx_x (customer_name) 
  ON PARTITION p_2020;  -- invalid syntax, no such feature
```

## What you *can* do to approximate per-partition control

If you genuinely need different indexing strategies per date range, the common workaround is:

1. **Separate physical tables** per partition-like range (e.g., `orders_2020`, `orders_2021`), each with its own custom indexes, unioned via a `VIEW` or merged manually.
2. Or, if using **partitioning + generated/virtual columns**, you can still only vary the *index definition* table-wide — not per partition.
3. **MyISAM** (deprecated, not recommended) and some other engines historically had more per-partition flexibility, but InnoDB (MySQL's default and effectively mandatory now) enforces uniform local indexing.

So functionally: you get isolated index structures per partition "for free" with any index you add, but you can't customize which indexes apply to which partitions.

If you tell me what you're trying to optimize for (e.g., faster queries on `p_before_2020` specifically, or reducing index overhead on rarely-queried old partitions), I can suggest whether partition pruning + a well-chosen composite index would already solve it without needing per-partition control.

    The ROLLBACK isnt working as you can see, this is because: 

The behavior you're seeing makes sense once you know how the `%%sql` magic (ipython-sql / jupysql) talks to MySQL.

**What's actually happening:**

Each `%%sql` cell just sends its statement over the connection, and by default that connection runs in **autocommit mode**. That means:

1. `START TRANSACTION;` runs
2. `UPDATE Accounts SET Balance = Balance - 1000 ...;` runs — and because autocommit is on, MySQL commits it immediately after execution (or the `mysqlconnector` driver commits between statements), regardless of the `START TRANSACTION` you issued
3. By the time `ROLLBACK;` runs, there's nothing left to roll back — the change was already committed

So the balance genuinely became 4000 permanently the moment the `UPDATE` executed. The `ROLLBACK` had no transaction to undo.

**Why this trips people up:** `START TRANSACTION` and the following statements need to be executed *within the same, non-autocommitting connection/session* for `ROLLBACK` to have any effect. If jupysql opens autocommit=True (which is the default for many SQLAlchemy MySQL connections), every individual statement is its own implicit transaction that commits right away.

**How to fix it:**

 Turn off autocommit for the sql magic:
```python
%config SqlMagic.autocommit = False
```
Put this in a cell before you run your transaction cells. Now statements won't be committed until you explicitly `COMMIT;` or until `ROLLBACK;` undoes them.

    Claude's Explanation:
    Good question — and yes, for raw `%%sql` cells (no Python connector), the safest way to see the *in-progress* state of a transaction is to keep the whole transaction inside **one cell**.

**Why:** jupysql executes each `%%sql` cell as its own call against the SQLAlchemy engine. Even though it usually reuses the same named connection across cells, the way the engine checks connections in/out of its pool between cell executions isn't guaranteed to preserve an *open, uncommitted* transaction across cell boundaries — especially depending on your jupysql/SQLAlchemy version. That's most likely why your `UPDATE` in one cell didn't show up in the `SELECT` in the next cell, even with `autocommit = False`.

**Fix — put it all in one `%%sql` cell:**

```sql
%%sql
START TRANSACTION;

UPDATE Accounts SET Balance = Balance - 1000 WHERE AccountID = 1;

SELECT * FROM Accounts;

ROLLBACK;

SELECT * FROM Accounts;
```

Since `%%sql` sends all statements in that cell sequentially over the *same* connection in one go, you'll correctly see:
- First `SELECT` → `4000` (inside the uncommitted transaction)
- Second `SELECT` → `5000` (after rollback)

14) Normalization

    
    Normalization in SQL is a systematic process of organizing the columns (attributes) and tables (relations) of a relational database to minimize data redundancy 
    and improve data integrity. In simpler terms, normalization helps ensure that each piece of data is stored in the best possible place, 
    reducing duplication and making it easier to maintain and update the database.

    Why Normalize?
    Reduce Data Redundancy: Storing the same data multiple times can waste space and cause inconsistencies if that data needs to change.
    Improve Data Integrity: Ensures that your data follows certain rules, preventing errors.
    Make Maintenance Easier: Updates, deletions, and insertions become simpler and more efficient when data is well-organized.

    Common Normal Forms
    The most commonly taught normal forms are:

    First Normal Form (1NF)
    Second Normal Form (2NF)
    Third Normal Form (3NF)

    (There are higher normal forms like 4NF and 5NF, but they are more specialized and not always required for most practical applications. 
    Typically, going up to 3NF or BCNF is sufficient.)


    1. First Normal Form (1NF)
    Definition & Purpose (1NF)
    1NF requires that every column holds only atomic (indivisible) values and that there are no repeating groups.
    Purpose: To ensure each field contains a single piece of data.

    Before 
    -- Non-1NF Table: contains multi-valued phone_numbers in a single column

    CREATE TABLE Students_Non1NF (
    student_id   INT,
    student_name VARCHAR(100),
    phone_numbers VARCHAR(100)  -- e.g., '123-4567,987-6543'
);

    -- Sample data insertion (non-atomic phone numbers)
    INSERT INTO Students_Non1NF (student_id, student_name, phone_numbers)
    VALUES (1, 'Alice', '123-4567,987-6543'),
           (2, 'Bob', '555-1212');

    After 

    -- Main Students table with atomic values
    CREATE TABLE Students (
        student_id   INT PRIMARY KEY,
        student_name VARCHAR(100)
    );

    -- Separate table for phone numbers, ensuring each phone number is atomic
    CREATE TABLE StudentPhones (
        student_id INT,
        phone      VARCHAR(20),
        PRIMARY KEY (student_id, phone),
        FOREIGN KEY (student_id) REFERENCES Students(student_id)
    );

    -- Insert students
    INSERT INTO Students (student_id, student_name)
    VALUES (1, 'Alice'),
        (2, 'Bob');

    -- Insert phone numbers (each phone number in its own row)
    INSERT INTO StudentPhones (student_id, phone)
    VALUES (1, '123-4567'),
        (1, '987-6543'),
        (2, '555-1212');


    Two phone number of a same guy has been netered in comma seperated form, which should not be the case as it is not atomic. No single column shall be repeated again as well. This defines NF1. Now let's implement it:

````python
%%sql
DROP TABLE IF EXISTS Students_Non1NF;
 CREATE TABLE Students_Non1NF (
    student_id   INT,
    student_name VARCHAR(100),
    phone_numbers VARCHAR(100)  -- e.g., '123-4567,987-6543'
);
````

````python
%%sql

INSERT INTO Students_Non1NF (student_id, student_name, phone_numbers)
    VALUES (1, 'Alice', '123-4567,987-6543'),
           (2, 'Bob', '555-1212');
````

````python
%%sql

SELECT * FROM Students_Non1NF;
````

    In order to avoid this we have split the phone number of students into seperate table, and use student_id as foreign key, and phone number and student_id as composite primary keys (so even if a student_id is repeated with another phone numbe , it still becomes unqiue).

````python
%%sql

-- Main Students table with atomic values
DROP TABLE IF EXISTS Students;
CREATE TABLE Students (
        student_id   INT PRIMARY KEY,
        student_name VARCHAR(100)
    );
````

````python
%%sql
-- Separate table for phone numbers, ensuring each phone number is atomic
CREATE TABLE StudentPhones (
    student_id INT,
    phone      VARCHAR(20),
    PRIMARY KEY (student_id, phone),
    FOREIGN KEY (student_id) REFERENCES Students(student_id)
);
````

    CREATE TABLE Courses_Non3NF (
        course_id        INT PRIMARY KEY,
        course_name      VARCHAR(100),
        instructor       VARCHAR(100),

    );
